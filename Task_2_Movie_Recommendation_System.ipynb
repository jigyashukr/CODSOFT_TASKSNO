# Simple Movie Recommendation System

This project builds a simple **content-based recommendation system** in Python. It recommends movies to users based on the genres of a movie they like.

The system uses **TF-IDF vectorization** and **cosine similarity** to find movies with similar genres.

## 1. Importing Libraries

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

## 2. Create the Movie Dataset

```python
movies = {
    "title": [
        "The Matrix",
        "Inception",
        "Fight Club",
        "Interstellar",
        "The Dark Knight",
        "Avengers: Endgame",
        "The Shawshank Redemption",
        "Titanic",
        "The Notebook",
        "Toy Story",
        "World War Z",
        "Finding Nemo",
        "Jurassic Park",
        "The Hangover",
        "John Wick"
    ],
    "genre": [
        "action science fiction",
        "action science fiction thriller",
        "action drama thriller",
        "science fiction drama adventure",
        "action crime drama",
        "action adventure science fiction",
        "crime drama",
        "romance drama",
        "romance drama",
        "animation comedy family",
        "action horror thriller science fiction",
        "animation adventure family",
        "science fiction adventure thriller",
        "comedy",
        "action crime thriller"
    ]
}

df = pd.DataFrame(movies)
df
```

## 3. Convert Genres into Numerical Features

```python
vectorizer = TfidfVectorizer()

genre_matrix = vectorizer.fit_transform(df["genre"])

print("Shape of the TF-IDF matrix:", genre_matrix.shape)
```

## 4. Calculate Movie Similarity

```python
similarity_matrix = cosine_similarity(genre_matrix)

print("Similarity matrix shape:", similarity_matrix.shape)
```

## 5. Create the Recommendation Function

The function takes a movie name as input, finds its similarity scores, sorts the movies from most similar to least similar, and returns recommendations.

```python
def recommend_movies(movie_title, num_recommendations=5):
    # Check whether the movie exists in the dataset
    if movie_title not in df["title"].values:
        return f"Sorry, '{movie_title}' is not available in the dataset."

    # Find the index of the selected movie
    movie_index = df.index[df["title"] == movie_title][0]

    # Get similarity scores for the selected movie
    similarity_scores = list(enumerate(similarity_matrix[movie_index]))

    # Sort movies by similarity score in descending order
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Remove the selected movie itself
    similarity_scores = similarity_scores[1:num_recommendations + 1]

    # Get movie names and similarity scores
    recommendations = []
    for index, score in similarity_scores:
        recommendations.append({
            "Movie": df.iloc[index]["title"],
            "Similarity Score": round(score, 2)
        })

    return pd.DataFrame(recommendations)
```

## 6. Test the Recommendation System

Let's ask the system for movies similar to **The Matrix**.

```python
recommend_movies("The Matrix")
```

## 7. Try Another Movie

```python
recommend_movies("Fight Club")
```

## 8. Interactive Recommendation System

Enter the name of a movie and the system will recommend similar movies.

```python
print("Available movies:")
for movie in df["title"]:
    print("-", movie)

print()
user_movie = input("Enter a movie you like: ")

recommendations = recommend_movies(user_movie)

print("\nRecommended movies:")
print(recommendations)
```

## 9. How This Recommendation System Works

1. A small movie dataset is created with movie titles and genres.
2. TF-IDF converts the genre text into numerical features.
3. Cosine similarity compares the movies.
4. The system finds movies with the highest similarity scores.
5. The most similar movies are recommended to the user.

This is an example of **content-based filtering** because recommendations are based on the characteristics (genres) of the items rather than ratings from other users.

## 10. Conclusion

The project demonstrates the basic working of a recommendation system using content-based filtering. Although the dataset is small, the same approach can be extended to larger movie, book, music, or product datasets.
