# Tic-Tac-Toe AI Agent

This project implements an AI agent that plays the classic **Tic-Tac-Toe** game against a human player.

The AI uses the **Minimax algorithm with Alpha-Beta Pruning** to make optimal decisions. With this strategy, the AI will never lose when it plays correctly.

## 1. Game Board

The Tic-Tac-Toe board contains 9 positions numbered from 1 to 9.

```python
board = [" " for _ in range(9)]

def display_board(board):
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()
```

## 2. Check for a Winner

There are 8 possible winning combinations: 3 rows, 3 columns, and 2 diagonals.

```python
def check_winner(board, player):
    winning_combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    return any(
        board[a] == board[b] == board[c] == player
        for a, b, c in winning_combinations
    )


def is_board_full(board):
    return " " not in board
```

## 3. Minimax with Alpha-Beta Pruning

The Minimax algorithm explores possible future moves and chooses the move that gives the AI the best possible outcome.

- **AI (`O`)** tries to maximize the score.
- **Human (`X`)** tries to minimize the score.
- Alpha-Beta Pruning skips branches that cannot affect the final decision, making the search more efficient.

```python
def minimax(board, depth, is_maximizing, alpha, beta):
    # Terminal states
    if check_winner(board, "O"):
        return 10 - depth

    if check_winner(board, "X"):
        return depth - 10

    if is_board_full(board):
        return 0

    # AI's turn: maximize the score
    if is_maximizing:
        best_score = float("-inf")

        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(board, depth + 1, False, alpha, beta)
                board[i] = " "

                best_score = max(best_score, score)
                alpha = max(alpha, best_score)

                # Alpha-Beta pruning
                if beta <= alpha:
                    break

        return best_score

    # Human's turn: minimize the score
    else:
        best_score = float("inf")

        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                score = minimax(board, depth + 1, True, alpha, beta)
                board[i] = " "

                best_score = min(best_score, score)
                beta = min(beta, best_score)

                # Alpha-Beta pruning
                if beta <= alpha:
                    break

        return best_score
```

## 4. Find the Best Move for the AI

The AI checks every available position and uses Minimax to determine which move has the best outcome.

```python
def best_ai_move(board):
    best_score = float("-inf")
    best_move = None

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(board, 0, False, float("-inf"), float("inf"))
            board[i] = " "

            if score > best_score:
                best_score = score
                best_move = i

    return best_move
```

## 5. Human Player Move

The human player enters a number from 1 to 9 to select an empty position.

```python
def human_move(board):
    while True:
        try:
            position = int(input("Enter your move (1-9): ")) - 1

            if position < 0 or position > 8:
                print("Please enter a number from 1 to 9.")
            elif board[position] != " ":
                print("That position is already occupied. Try again.")
            else:
                board[position] = "X"
                break

        except ValueError:
            print("Please enter a valid number.")
```

## 6. Play Tic-Tac-Toe

The human player uses `X` and the AI uses `O`. The human goes first.

```python
def play_game():
    board = [" " for _ in range(9)]

    print("Welcome to Tic-Tac-Toe!")
    print("You are X and the AI is O.")
    print("Choose positions using the numbers below:")
    print()
    print(" 1 | 2 | 3 ")
    print("---+---+---")
    print(" 4 | 5 | 6 ")
    print("---+---+---")
    print(" 7 | 8 | 9 ")

    while True:
        # Human turn
        human_move(board)
        display_board(board)

        if check_winner(board, "X"):
            print("Congratulations! You won!")
            break

        if is_board_full(board):
            print("It's a draw!")
            break

        # AI turn
        print("AI is thinking...")
        ai_move = best_ai_move(board)
        board[ai_move] = "O"
        display_board(board)

        if check_winner(board, "O"):
            print("AI wins! Better luck next time.")
            break

        if is_board_full(board):
            print("It's a draw!")
            break
```

## 7. Start the Game

Run the cell below to play against the AI.

```python
play_game()
```

## 8. How the AI Works

The AI uses the following process:

1. It looks at all currently available moves.
2. For each move, it simulates possible future moves.
3. Minimax assigns scores to possible game outcomes.
4. The AI chooses the move with the highest score.
5. Alpha-Beta Pruning removes branches that cannot improve the result.

Because Tic-Tac-Toe has a small game tree, Minimax can search the possible outcomes deeply enough to play optimally.
