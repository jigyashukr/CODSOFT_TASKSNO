# Rule-Based Chatbot

A simple chatbot built using Python `if-elif-else` statements and keyword pattern matching. The chatbot identifies common user queries and provides predefined responses.

## 1. Importing Library

```python
from datetime import datetime
```

## 2. Define the Chatbot Response Function

```python
def chatbot_response(user_input):
    # Converting input to lowercase and remove extra spaces
    user_input = user_input.lower().strip()

    # Greeting
    if any(word in user_input.split() for word in ["hello", "hi", "hey"]):
        return "Hello! How can I help you?"

    # Asking the chatbot's name
    elif "your name" in user_input or "who are you" in user_input:
        return "I am SMASH, a simple rule-based chatbot."

    # Asking how the chatbot is doing
    elif "how are you" in user_input:
        return "I'm doing great! Thanks for asking."

    # Asking about chatbot capabilities
    elif "what can you do" in user_input or "help" in user_input:
        return "I can answer simple questions using predefined rules."

    # Asking about Python
    elif "python" in user_input:
        return "Python is a popular programming language used for AI, ML, web development, and data analysis."

    # Asking about Machine Learning
    elif "machine learning" in user_input or "ml" in user_input:
        return "Machine Learning allows computers to learn patterns from data and make predictions."

    # Asking about Artificial Intelligence
    elif "artificial intelligence" in user_input or "ai" in user_input:
        return "Artificial Intelligence is the technology that enables machines to learn, reason, and perform tasks that normally require human intelligence."

    # Asking for the current time
    elif "time" in user_input:
        current_time = datetime.now().strftime("%H:%M:%S")
        return f"The current time is {current_time}."

    # Goodbye
    elif any(word in user_input for word in ["bye", "goodbye", "exit", "quit"]):
        return "Goodbye! Have a great day!"

    # Unknown input
    else:
        return "Sorry, I don't understand that. Please try asking something else."
```

## 3. Test the Chatbot

We can test the chatbot with a few predefined inputs.

```python
test_inputs = [
    "Hello",
    "What is your name?",
    "How are you?",
    "What is Python?",
    "What is machine learning?",
    "What is ai?",
    "What time is it?",
    "Bye"
]

for user_input in test_inputs:
    print("You:", user_input)
    print("Chatbot:", chatbot_response(user_input))
    print()
```

## 4. Interactive Chatbot

Run this cell to chat with the bot. Type `bye`, `exit`, or `quit` to end the conversation.

```python
print("Chatbot: Hello! I am a simple rule-based chatbot.")
print("Chatbot: Type 'bye' to end the conversation.\n")

while True:
    user_input = input("You: ")
    response = chatbot_response(user_input)
    print("Chatbot:", response)

    if user_input.lower().strip() in ["bye", "goodbye", "exit", "quit"]:
        break
```

## 5. Conclusion

This project demonstrates a basic form of Natural Language Processing (NLP) using rule-based pattern matching. The chatbot does not use machine learning; instead, it identifies keywords and phrases in the user's input and selects an appropriate predefined response using `if-elif-else` statements.
